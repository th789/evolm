```python
import shutil
from itertools import product
import re
```

```python
##args
sft_dataset = 'medmcqa' #options: ['hellaswag', 'medmcqa', 'pubmedqa', 'mmluprocot'], reference = metamathqa, so it's NOT an option
```

### write config files for SFT, no weight decay (aka weight decay = 0)

```python
model_size_lst = ['llama-1B-20BT', 'llama-0.5B-10BT']
wd_pretrain_lst = [0.0001, 0.001, 0.01, 0.1, 0.5, 1.0, 1.5, 3.0, 10.0]
ft_seed_lst = [8107, 5701]
```

```python
n_newconfig_files_created = 0

for model_size, wd_pretrain, ft_seed in product(model_size_lst, wd_pretrain_lst, ft_seed_lst):
    
    #make a copy of existing config file for sft_dataset
    print(f"Creating config file for: model {model_size}, wd_pretrain: {wd_pretrain}")
    og_config_file_path = f'config_hub/custom_configs/ft_{sft_dataset}/{model_size}-weightdecay{wd_pretrain}-seed42-{sft_dataset}.yaml'

    new_config_file_path = f'config_hub/custom_configs/ft_{sft_dataset}/{model_size}-weightdecay{wd_pretrain}-seed42-{sft_dataset}-ftseed{ft_seed}.yaml'
    shutil.copyfile(og_config_file_path, new_config_file_path)

    ### make edits to the new config file
    with open(new_config_file_path, "r", encoding="utf-8") as f:
        text = f.read()

        #add seed as an argument to config file
        #find "do_eval: false" string, add "seed: ft_seed" after it
        old_str = "do_eval: false"
        new_str = old_str + f"\nseed: {ft_seed}"
        text = text.replace(old_str, new_str)

        #add ft_seed to output_dir + run_name
        #find "output_dir:" string, add "-ftseed{ft_seed}" to the end of the line
        text = re.sub(
            r"^output_dir:\s*(.+)$",
            rf"output_dir: \1-ftseed{ft_seed}",
            text,
            flags=re.MULTILINE,
        )

        #find "run_name:" string, add "-ftseed{ft_seed}" to the end of the line
        text = re.sub(
            r"^run_name:\s*(.+)$",
            rf"run_name: \1-ftseed{ft_seed}",
            text,
            flags=re.MULTILINE,
        )

        #write new config file
        with open(new_config_file_path, "w", encoding="utf-8") as f:
            f.write(text)
        
    n_newconfig_files_created += 1



print('n_newconfig_files_created: ', n_newconfig_files_created)
print("Complete!")
```
