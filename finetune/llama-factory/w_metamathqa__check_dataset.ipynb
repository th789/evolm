```python
import pandas as pd
```

```python
#compare with response length of metamathqa dataset

#load metamathqa dataset, json file
df = pd.read_json("data/metamathqa.json")
df.head()
```

```python
# check length of response
df['response_length'] = df['output'].apply(len)

#print stats
print(df['response_length'].describe())
```

```python
#get index of row with max length
idx_of_row_w_max_length = df['response_length'].idxmax()
print(idx_of_row_w_max_length)


#print row with max length
print('Question:')
print(df.iloc[idx_of_row_w_max_length]['instruction'])

print('\nResponse:')
print(df.iloc[idx_of_row_w_max_length]['output'])

```
