start with config files used in exp01 (finetune 1B models that were *pretrained* with varying weight decay)

make new config files that vary weight_decay during *finetuning*

```python
import shutil
from itertools import product
```

```python
#full combinations
# wd_pretrain_lst = [0.0001, 0.001, 0.01, 0.1, 1.0]
# wd_ft_lst = [0.01, 0.1, 1.0]
#new experiments
wd_pretrain_lst = [3.0, 10.0]
wd_ft_lst = [0.01, 0.1, 1.0]

#small test
# wd_pretrain_lst = [0.1]
# wd_ft_lst = [0.01]
```

```python
#count # of new config files created
n_newconfig_files_created = 0

for wd_pretrain, wd_ft in product(wd_pretrain_lst, wd_ft_lst):
    print(f"Creating config file for wd_pretrain: {wd_pretrain}, wd_ft: {wd_ft}")
    #copy config file used to finetune the pretrained model
    og_config_file_path = f'config_hub/custom_configs/ft_metamathqa/llama-1B-20BT-weightdecay{wd_pretrain}-seed42-metamathqa.yaml'
    new_config_file_path = f'config_hub/custom_configs/ft_metamathqa/vary_wd_during_ft/llama-1B-20BT-weightdecay{wd_pretrain}-seed42-metamathqa-ftweightdecay{wd_ft}.yaml'
    shutil.copyfile(og_config_file_path, new_config_file_path)


    ### make edits to the new config file

    #load the new config file as a string
    with open(new_config_file_path, "r", encoding="utf-8") as f:
        text = f.read()

    #insert weight_decay argument in new line after do_eval argument
    old_str = "do_eval: false"
    insert_str = f"weight_decay: {wd_ft}"
    new_str = f"{old_str}\n{insert_str}" 
    text = text.replace(old_str, new_str)

    #rename output_dir: change folder + add ftweightdecay{wd_ft} to the end of the string
    old_str = f"output_dir: llamafactory_out/llama-1B-20BT-weightdecay{wd_pretrain}-seed42-metamathqa"
    new_str = f"output_dir: llamafactory_out/vary_wd_during_ft/llama-1B-20BT-weightdecay{wd_pretrain}-seed42-metamathqa-ftweightdecay{wd_ft}"
    text = text.replace(old_str, new_str)

    #rename run_name for wandb logging
    old_str = f"run_name: llama-1B-20BT-weightdecay{wd_pretrain}-seed42-metamathqa"
    new_str = f"run_name: llama-1B-20BT-weightdecay{wd_pretrain}-seed42-metamathqa-ftweightdecay{wd_ft}"
    text = text.replace(old_str, new_str)

    #write new config file
    with open(new_config_file_path, "w", encoding="utf-8") as f:
        f.write(text)

    n_newconfig_files_created += 1

print(n_newconfig_files_created)
print("Complete!")
```
