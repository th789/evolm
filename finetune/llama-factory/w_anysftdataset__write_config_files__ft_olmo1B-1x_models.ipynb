```python
import shutil
from itertools import product
```

```python
##args -- write new config files with this setup
sft_dataset = 'simplescaling' #options: ['metamathqa', 'medmcqa', 'hellaswag',  'pubmedqa', 'mmluprocot',  'race', 'simplescaling'] #did not do hellaswag
```

### write config files for SFT, based on olmo model configs

```python
model_size_lst = ['olmo-1B-30BT'] #creating new config files for this model size
wd_pretrain_lst = [0.1, 1.0] #[0.1, 1.0]

model_path_dict = {
    'olmo-1B-30BT-weightdecay0.1': 'sbordt/OLMo-2-1B-1x',
    'olmo-1B-30BT-weightdecay1.0': 'sbordt/OLMo-2-1B-1x-WD1',
}


#uses the config file of this model for reference -- first copied then edited
model_size_reference = 'olmo-1B-210BT' 
model_path_dict_reference = {
    'olmo-1B-210BT-weightdecay0.1': 'sbordt/OLMo-2-1B-Decayed-Early',
    'olmo-1B-210BT-weightdecay1.0': 'sbordt/OLMo-2-1B-WD1',
}
```

```python
n_newconfig_files_created = 0

for model_size, wd_pretrain in product(model_size_lst, wd_pretrain_lst):

    #make a copy of config file for metamathqa
    print(f"Creating config file for: model {model_size}, wd_pretrain {wd_pretrain}, sft_dataset {sft_dataset}")
    og_config_file_path = f'config_hub/custom_configs/ft_{sft_dataset}/{model_size_reference}-weightdecay{wd_pretrain}-{sft_dataset}.yaml'
    new_config_file_path = f'config_hub/custom_configs/ft_{sft_dataset}/{model_size}-weightdecay{wd_pretrain}-{sft_dataset}.yaml'
    shutil.copyfile(og_config_file_path, new_config_file_path)

    ### make edits to the new config file
    with open(new_config_file_path, "r", encoding="utf-8") as f:
        text = f.read()

        #changes these variables in config file:
            # model_name_or_path: olmo-1B-30BT has different directory than olmo-1B-210BT
            # output_dir
            # run_name

        #change model_name_or_path
        key_reference = f"{model_size_reference}-weightdecay{wd_pretrain}"
        old_str = model_path_dict_reference[key_reference]
        
        key = f"{model_size}-weightdecay{wd_pretrain}"
        new_str = f"{model_path_dict[key]}"
        text = text.replace(old_str, new_str)

        #change other appearances of model_size -- effectively changing output_dir and run_name
        text = text.replace(model_size_reference, model_size)
        
        #write new config file
        with open(new_config_file_path, "w", encoding="utf-8") as f:
            f.write(text)

    n_newconfig_files_created += 1

print(n_newconfig_files_created)
print("Complete!")
```
