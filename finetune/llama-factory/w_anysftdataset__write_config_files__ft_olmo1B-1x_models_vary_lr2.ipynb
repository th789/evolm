```python
#for olmo-1x models, wd=0.1, 0.6, 1.0 have varying learning rates
    #each wd uses default lr=4e-4 (not in model name) and lr=2e-4, lr=8e-4



###what this script does:
    #start with config file for olmo-1B-30BT-weightdecay{wd}-{sft_dataset}.yaml -- aka config file for model with default lr
    #change model path
    #change output dir
    #change run name
    #(don't need to change sft dataset since it's the same as the starting config)
```

```python
import shutil
from itertools import product
```

```python
##args -- write new config files with this setup

#arguments that change -- re-run notebook for each
wd_pt = 1.0 #options: 0.1, 0.6, 1.0

#arguments that do not change
model_size = 'olmo-1B-30BT'
lr_lst = ['2e-4', '8e-4']
sft_dataset_lst = ['metamathqa', 'medmcqa', 'pubmedqa', 'mmluprocot',  'race', 'simplescaling']




hf_name_defaultlr_model = {
    'olmo-1B-30BT-weightdecay0.1': 'sbordt/OLMo-2-1B-1x',
    'olmo-1B-30BT-weightdecay0.3': 'sbordt/OLMo-2-1B-1x-WD03',
    'olmo-1B-30BT-weightdecay0.6': 'sbordt/OLMo-2-1B-1x-WD06',
    'olmo-1B-30BT-weightdecay1.0': 'sbordt/OLMo-2-1B-1x-WD1',
}


hf_name_varyinglr_model = {
    'olmo-1B-30BT-weightdecay0.1-lr2e-4': 'sbordt/OLMo-2-1B-1x-WD01-LR02',
    'olmo-1B-30BT-weightdecay0.1-lr8e-4': 'sbordt/OLMo-2-1B-1x-WD01-LR08',
    'olmo-1B-30BT-weightdecay0.3-lr2e-4': 'sbordt/OLMo-2-1B-1x-WD03-LR02',
    'olmo-1B-30BT-weightdecay0.3-lr8e-4': 'sbordt/OLMo-2-1B-1x-WD03-LR08',
    'olmo-1B-30BT-weightdecay0.6-lr2e-4': 'sbordt/OLMo-2-1B-1x-WD06-LR02',
    'olmo-1B-30BT-weightdecay0.6-lr8e-4': 'sbordt/OLMo-2-1B-1x-WD06-LR08',
    'olmo-1B-30BT-weightdecay1.0-lr2e-4': 'sbordt/OLMo-2-1B-1x-WD1-LR02',
    'olmo-1B-30BT-weightdecay1.0-lr8e-4': 'sbordt/OLMo-2-1B-1x-WD1-LR08',
}
```

```python
n_newconfig_files_created = 0


for lr in lr_lst:
    print(f"Creating config file for: {model_size}, wd_pretrain {wd_pt}, lr {lr}")

    for sft_dataset in sft_dataset_lst:
        print(f"   sft_dataset {sft_dataset}")

        #make a copy of config file for olmo-1B-wd{wd}-defaultlr model
        og_config_file_path = f'config_hub/custom_configs/ft_{sft_dataset}/{model_size}-weightdecay{wd_pt}-{sft_dataset}.yaml'
        new_config_file_path = f'config_hub/custom_configs/ft_{sft_dataset}/{model_size}-weightdecay{wd_pt}-lr{lr}-{sft_dataset}.yaml'
        shutil.copyfile(og_config_file_path, new_config_file_path)

        ### make edits to the new config file
        with open(new_config_file_path, "r", encoding="utf-8") as f:
            text = f.read()

            #changes these variables in config file:
                # model_name_or_path: olmo-1B-30BT has different directory than olmo-1B-210BT
                # output_dir
                # run_name

            #change model_name_or_path
            key_reference_model = f"{model_size}-weightdecay{wd_pt}"
            old_str = hf_name_defaultlr_model[key_reference_model]
            
            key_new_model = f"{model_size}-weightdecay{wd_pt}-lr{lr}"
            new_str = f"{hf_name_varyinglr_model[key_new_model]}"
            text = text.replace(old_str, new_str)

            #change other appearances of model_size -- effectively changing output_dir and run_name
            text = text.replace(key_reference_model, key_new_model)
            
            #write new config file
            with open(new_config_file_path, "w", encoding="utf-8") as f:
                f.write(text)

        n_newconfig_files_created += 1

print(n_newconfig_files_created)
print("Complete!")
```
