```python
###what this script does:
    #start with config file for olmo-1B-30BT-weightdecay0.1-{sft_dataset}.yaml
    # for each sft
        #change model path: from sbordt/OLMo-2-1B-1x to sbordt/OLMo-2-1B-1x-WD03
        #change output dir
        #change run name
        #(don't need to change sft dataset since it's the same as the starting config)
```

```python
import shutil
from itertools import product
```

```python
##args -- write new config files with this setup
sft_dataset_lst = ['metamathqa', 'medmcqa', 'pubmedqa', 'mmluprocot',  'race', 'simplescaling'] #options: ['metamathqa', 'medmcqa', 'hellaswag',  'pubmedqa', 'mmluprocot',  'race', 'simplescaling'] #did not do hellaswag
```

### write config files for SFT, based on olmo model configs

```python
model = 'olmo-1B-30BT-weightdecay0.3' #creating new config files for this model size
model_reference = 'olmo-1B-30BT-weightdecay0.1' #uses the config file of this model for reference -- first copied then edited

model_path_dict = {
    'olmo-1B-30BT-weightdecay0.1': 'sbordt/OLMo-2-1B-1x',
    'olmo-1B-30BT-weightdecay0.3': 'sbordt/OLMo-2-1B-1x-WD03',
    'olmo-1B-30BT-weightdecay1.0': 'sbordt/OLMo-2-1B-1x-WD1',
}
```

```python
old_str = model_path_dict[model_reference]
new_str = model_path_dict[model]
```

```python
n_newconfig_files_created = 0

for sft_dataset in sft_dataset_lst:

    #make a copy of model_reference's config file for this sft_dataset
    print(f"Creating config file for: model {model}, sft_dataset {sft_dataset}")
    og_config_file_path = f'config_hub/custom_configs/ft_{sft_dataset}/{model_reference}-{sft_dataset}.yaml'
    new_config_file_path = f'config_hub/custom_configs/ft_{sft_dataset}/{model}-{sft_dataset}.yaml'
    shutil.copyfile(og_config_file_path, new_config_file_path)

    ### make edits to the new config file
    with open(new_config_file_path, "r", encoding="utf-8") as f:
        text = f.read()

        #changes these variables in config file:
            # model_name_or_path: olmo-1B-30BT has different directory than olmo-1B-210BT
            # output_dir
            # run_name

        #change model_name_or_path
        old_str = model_path_dict[model_reference]
        new_str = model_path_dict[model]
        text = text.replace(old_str, new_str)

        #change other appearances of model_size -- effectively changing output_dir and run_name
        text = text.replace(model_reference, model)
        
        #write new config file
        with open(new_config_file_path, "w", encoding="utf-8") as f:
            f.write(text)

    n_newconfig_files_created += 1

print(n_newconfig_files_created)
print("Complete!")
```
