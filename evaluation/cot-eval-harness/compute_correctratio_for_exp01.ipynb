```python
import os
import json
import numpy as np
import itertools
```

```python
#arguments
model_size_lst = ['1B-20BT', '0.5B-10BT']
weight_decay_lst = [0.0001, 0.001, 0.01, 0.1, 1.0]
generation_type_lst = ['n16', 'greedy']
dataset_lst = [
    'BoardgameQA500', 'CRUXEval', 
    'GSM8KPlatinum', 'MATHHard', 'MATHLevel1', 'MATHLevel2', 'MATHLevel3', 'MATHLevel4', 
    'StrategyQA500', 'TabMWP'
    ]

combinations = itertools.product(model_size_lst, weight_decay_lst, generation_type_lst, dataset_lst)
```

```python
counter = 0
for model_size, weight_decay, generation_type, dataset in combinations:
    counter += 1
    print(f"{counter}. Processing model size = {model_size}, weight decay = {weight_decay}, generation type = {generation_type}, dataset = {dataset}")

    #### load grader_results
    eval_results_folder = "eval_output/complete--with-correctratio" #!!!now folder is named "complete"
    model_name = f"llama-{model_size}-weightdecay{weight_decay}-seed42-metamathqa--{generation_type}"
    exam_out_dir = os.path.join(eval_results_folder, model_name, dataset)

    grader_results_file = os.path.join(exam_out_dir, "chunk_0-1", "grader_results.json")

    with open(grader_results_file, "r") as f:
        grader_results = json.load(f)
    print(f"Loaded grader results from {grader_results_file}")


    ##### calculate correct ratio
    #inititate list to store correct_answer_ratios
    correct_ratio_lst = []

    #collect correct_answer_ratio for questions where correct_answer_existence is True
    n_questions = len(grader_results['updated_examinee_results'])

    for i in range(n_questions):
        correct_answer_existence = grader_results['updated_examinee_results'][i]['correct_answer_existence']
        if correct_answer_existence:
            correct_answer_ratio = grader_results['updated_examinee_results'][i]['correct_answer_ratio']
            correct_ratio_lst.append(correct_answer_ratio)

    #compute correct_ratio_at_k
    correct_ratio_at_k_acc = np.mean(correct_ratio_lst)


    #### re-write the grader_results_overview.json file
    #load the grader_results_overview.json file
    grader_results_overview_file = os.path.join(exam_out_dir, "chunk_0-1", "grader_results--overview.json")
    with open(grader_results_overview_file, "r") as f:
        grader_results_overview = json.load(f)

    #add correct_ratio_at_k
    grader_results_overview["correct_ratio_at_k_acc"] = correct_ratio_at_k_acc

    #save the grader_results_overview.json file --> overwrites original file
    with open(grader_results_overview_file, "w") as f:
        json.dump(grader_results_overview, f, indent=4)
            
```
