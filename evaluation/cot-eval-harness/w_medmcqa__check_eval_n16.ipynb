```python
import json
import re
from tqdm import tqdm
from collections import Counter
import numpy as np
```

```python
#load collected responses for medmcqa
wd = 1.0
model_name = f"llama-1B-20BT-weightdecay{wd}-seed42-medmcqa"

#load collected responses for medmcqa
with open(f"eval_output/complete/medmcqa/{model_name}--n16/medmcqa/chunk_0-1/examinee_results.json", "r", encoding="utf-8") as f:
    responses_medmcqa_n16 = json.load(f)
```

```python
def extract_from_answer_colon(text):
    match = re.search(r'Answer:\s*([A-Za-z])', text, re.IGNORECASE)
    #if "Answer: X" or "answer: X" is in the text, return X
    if match:
        return match.group(1)
    return None
```

```python
numerator_pass_at_k = 0
numerator_majority_at_k = 0
numerator_correct_ratio_at_k = 0

n_responses_w_at_least_one_correct = 0


for i in tqdm(range(len(responses_medmcqa_n16))):
    response = responses_medmcqa_n16[i]
    correct_answer = response['gt_answer']

    #examine each of the 16 generations
    extracted_answers = []

    for generation_idx in range(16):
        generation = response['model_responses'][generation_idx]['text']
        extracted_answer = extract_from_answer_colon(generation)
        extracted_answers.append(extracted_answer)
        if extracted_answer is None:
            print(f'Question {i}, generation {generation_idx} has extracted answer = None')
            # print(f'generation: {generation}')

    extracted_answers_correct = [a == correct_answer for a in extracted_answers]

    ### pass at k
    if sum(extracted_answers_correct) > 0: #at least one generation is correct (True) in extracted_answers_correct
        numerator_pass_at_k += 1

    ### majority at k
    #get most common answer in extracted_answers
    most_common_letter = Counter(extracted_answers).most_common(1)[0][0]

    # check if it is correct -- if correct, add 1 to numerator_majority_at_k
    if most_common_letter == correct_answer:
        numerator_majority_at_k += 1

    ### correct ratio at k
    if sum(extracted_answers_correct) > 0: #at least one generation is correct (True) in extracted_answers_correct
        correct_ratio_for_this_response = np.mean(extracted_answers_correct)
        numerator_correct_ratio_at_k += correct_ratio_for_this_response
        n_responses_w_at_least_one_correct += 1
```
