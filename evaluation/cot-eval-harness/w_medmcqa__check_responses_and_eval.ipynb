### check response correctness by eye

```python
import re
```

```python
def extract_from_answer_colon(text):
    match = re.search(r'Answer:\s*([A-Za-z])', text, re.IGNORECASE)
    #if "Answer: X" or "answer: X" is in the text, return X
    if match:
        return match.group(1)
    return None
```

```python
n_correct = 0

for q in range(len(responses_medmcqa_greedy)):
    print('------', q, '------')
    correct_answer = responses_medmcqa_greedy[q]['gt_answer']
    model_answer = extract_from_answer_colon(responses_medmcqa_greedy[q]['model_responses'][0]['text'])
    print('Correct answer:', correct_answer)
    print('Model answer:', model_answer)

    if correct_answer == model_answer:
        n_correct += 1
```

```python
print('Accuracy:', n_correct / len(responses_medmcqa_greedy))
print('Number correct:', n_correct)
```
